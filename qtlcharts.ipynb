```r
library(qtlcharts)
library(htmlwidgets)
library(IRdisplay)
library(glue)
library(qtl)
```

```r
if(!dir.exists("html")) dir.create("html")
```

```r
file <- "html/iplot_demo.html"
height <- 500
width <- 700

set.seed(20260531)
x <- rnorm(100)
y <- x+rnorm(100)

h <- iplot(x,y, chartOpts=list(height=height, width=width))

saveWidget(h, file, selfcontained=TRUE)
html <- glue('<iframe src="{file}" height="{height+100}px" width="{width+100}px"></iframe>')
display_html(html)
```

```r
data(hyper)

file <- "html/iplotMap_demo.html"
height <- 700
width <- 850

h <- iplotMap(pull.map(hyper), chartOpts=list(height=height, width=width))

saveWidget(h, file, selfcontained=TRUE)
html <- glue('<iframe src="{file}" height="{height+100}px" width="{width+100}px"></iframe>')
display_html(html)
```

```r
data(grav)

file <- "html/iboxplot_demo.html"
height <- 680
width <- 850

h <- iboxplot(t(grav$pheno), breaks=51, orderByMedian=FALSE, 
             chartOpts=list(xlab="Time index", ylab="Root tip angle (degrees)",
                           height=height, width=width))

saveWidget(h, file, selfcontained=TRUE)
html <- glue('<iframe src="{file}" height="{height+100}px" width="{width+100}px"></iframe>')
display_html(html)
```

```r
file <- "html/iheatmap_demo.html"
height <- 650
width <- 850

n <- 101                                                                                                                                                      
x <- y <- seq(-2, 2, len=n)                                                                                                                                   
z <- matrix(ncol=n, nrow=n)                                                                                                                                   
for(i in seq(along=x))                                                                                                                                        
    for(j in seq(along=y))                                                                                                                                    
        z[i,j] <- x[i]*y[j]*exp(-x[i]^2 - y[j]^2)                                                                                                             
h <- iheatmap(z, x, y, chartOpts=list(height=height, width=width)) 

saveWidget(h, file, selfcontained=TRUE)
html <- glue('<iframe src="{file}" height="{height+100}px" width="{width+100}px"></iframe>')
display_html(html)
```

```r
file <- "html/iplotCorr_demo.html"
height <- 400
width <- 850

h <- iplotCorr(grav$pheno[, seq(1, ncol(grav$pheno), by=5)],
              chartOpts=list(height=height, width=width))

saveWidget(h, file, selfcontained=TRUE)
html <- glue('<iframe src="{file}" height="{height+100}px" width="{width+100}px"></iframe>')
display_html(html)
```

```r
file <- "html/iplotCurves_demo.html"
height <- 650
width <- 850

times <- as.numeric(sub("T", "", colnames(grav$pheno)))/60
h <- iplotCurves(grav$pheno, times,
                 grav$pheno[,c("T30", "T240")],
                 grav$pheno[,c("T240", "T480")],
                 chartOpts=list(curves_xlab="Time", curves_ylab="Root tip angle",
                                scat1_xlab="Angle at 30 min", scat1_ylab="Angle at 4 hrs",
                                scat2_xlab="Angle at 4 hrs", scat2_ylab="Angle at 8 hrs",
                               height=height, width=width))

saveWidget(h, file, selfcontained=TRUE)
html <- glue('<iframe src="{file}" height="{height+100}px" width="{width+100}px"></iframe>')
display_html(html)
```

```r
set.seed(69891250)                                                                                                                                            
p <- 500                                                                                                                                                      
n <- 300                                                                                                                                                      
SD <- runif(p, 1, 5)                                                                                                                                          
r <- runif(p, -1, 1)                                                                                                                                          
scat2 <- vector("list", p)                                                                                                                                    
for(i in 1:p)                                                                                                                                                 
   scat2[[i]] <- matrix(rnorm(2*n), ncol=2) %*% chol(SD[i]^2*matrix(c(1, r[i], r[i], 1), ncol=2))                                                             
scat1 <- cbind(SD=SD, r=r)   

width <- 800
height <- 400
file <- "html/scat2scat_demo.html"

h <- scat2scat(scat1, scat2, chartOpts=list(width=width, height=height))

saveWidget(h, file, selfcontained=TRUE)
html <- glue('<iframe src="{file}" height="{height+100}px" width="{width+100}px"></iframe>')
display_html(html)
```
